```python
import statistics
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv('/content/fitness_tracker.csv')
print("fitness dataset")
display(df.head())
member_names = df['user_id'].astype(str).tolist()
ages = df['age'].tolist()
workout_hours = df['workout_duration_minutes'].tolist()
calories_burned = df['calories_burned'].tolist()
average_workout = df['workout_duration_minutes'].mean()
print("average workout duration (minutes):", average_workout)

biggest_calories = df['calories_burned'].max()
print("biggest calories burned:", biggest_calories)

smallest_calories = df['calories_burned'].min()
print("smallest calories burned:", smallest_calories)

median_calories = df['calories_burned'].median()
print("median calories burned:", median_calories)

std_dev = df['calories_burned'].std()
print("standard deviation of calories burned:", std_dev)
plt.figure(figsize=(14, 6))
plt.bar(member_names, calories_burned)
plt.xlabel("User ID")
plt.ylabel("Calories Burned")
plt.title("Fitness Analysis (Calories Burned by User ID - Bar Chart)")
plt.show()
plt.figure(figsize=(14, 6))
plt.scatter(member_names, calories_burned)
plt.xlabel("User ID")
plt.ylabel("Calories Burned")
plt.title("Fitness Analysis (Calories Burned by User ID - Scatter Plot)")
plt.show()
```
